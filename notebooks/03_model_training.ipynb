# Treinamento e avaliação do modelo

### Carregar dados processados e artefatos

```python

import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix


# Como estamos na pasta 'notebooks', voltamos um nível com '../'
X_train = joblib.load('../data/processed/X_train_resampled.pkl')
y_train = joblib.load('../data/processed/y_train_resampled.pkl')
X_test = joblib.load('../data/processed/X_test_processed.pkl')
y_test = joblib.load('../data/processed/y_test.pkl')

print(f"Dados carregados com sucesso!")
print(f"Treino: {X_train.shape} | Teste: {X_test.shape}")
```

### Treinamento do modelo

1. Instanciar: Você "chama" o modelo vazio
2. Fit (Treinar): Você passa os dados de estudo para ele aprender (.fit())
3. Predict (Prever): Você passa os dados cegos (teste) para ele tentar adivinhar (.predict())


### Treinando com Regressão Logística e estabelecendo baseline

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# 1. Instanciar o modelo
modelo_lr = LogisticRegression(random_state=42, max_iter=1000)

# 2. Treinar (Fit) - ATENÇÃO: Usamos os dados que passaram pelo SMOTE!
modelo_lr.fit(X_train, y_train)

# 3. Prever (Predict) - ATENÇÃO: Usamos o teste, que NUNCA viu o SMOTE!
y_pred = modelo_lr.predict(X_test)

# --- AVALIAÇÃO BASELINE ---
print("Matriz de Confusão:")
print(confusion_matrix(y_test, y_pred))
print("\nRelatório de Classificação (Precision, Recall, F1):")
print(classification_report(y_test, y_pred))
```

### Treinando com Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import classification_report, confusion_matrix

# 1. Instanciar o modelo
modelo_lr = RandomForestClassifier(random_state=42, n_estimators=100)

# 2. Treinar (Fit) - ATENÇÃO: Usamos os dados que passaram pelo SMOTE!
modelo_lr.fit(X_train, y_train)

# 3. Prever (Predict) - ATENÇÃO: Usamos o teste, que NUNCA viu o SMOTE!
y_pred = modelo_lr.predict(X_test)

# --- AVALIAÇÃO RANDOM FOREST ---
print("Matriz de Confusão:")
print(confusion_matrix(y_test, y_pred))
print("\nRelatório de Classificação (Precision, Recall, F1):")
print(classification_report(y_test, y_pred))
```

```python
from xgboost import XGBClassifier

# 1. Instanciar (n_estimators é o número de árvores na sequência)
modelo_xgb = XGBClassifier(
    random_state=42,
    n_estimators=100,
    use_label_encoder=False,
    eval_metric='logloss'
)

# 2. Fit
modelo_xgb.fit(X_train, y_train)

# 3. Predict
y_pred = modelo_xgb.predict(X_test)

# --- AVALIAÇÃO XGBOOST ---
print("Matriz de Confusão:")
print(confusion_matrix(y_test, y_pred))
print("\nRelatório de Classificação (Precision, Recall, F1):")
print(classification_report(y_test, y_pred))
```
